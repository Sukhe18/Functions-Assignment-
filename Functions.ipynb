Q.4 What is the purpose of the `return` statement in a Python function?

- The most common use is to make the function output a result that can be used elsewhere in the program.

When the Python interpreter encounters a return statement, the function immediately stops executing, and control is passed back to the point where the function was called. Any code following return inside the function will not be executed (it is often referred to as "unreachable code").

Q.5 What are iterators in Python and how do they differ from iterables?

- An iterable is any Python object capable of returning its members one at a time. In simpler terms, it's anything you can "iterate" over using a for loop.

Key Characteristics of an Iterable:

- Protocol: An iterable must have the special method __iter__(). When called, this method returns an iterator object.

- Examples: Lists, tuples, strings, sets, dictionaries, and generators are all iterables.

Iterators:

An iterator is an object that represents a stream of data. It is the object that actually does the act of iterating, keeping track of the current position and fetching the next element.

Key Characteristics of an Iterator:

Protocol: An iterator must implement two special methods:

1. __iter__(): Returns the iterator object itself. This is what makes an iterator also an iterable (though typically not one you'd iterate over directly).

2. __next__(): Returns the next item from the sequence. If there are no more items, it must raise the StopIteration exception, which signals to the for loop to terminate.

Q.6 Explain the concept of generators in Python and how they are defined.


- Unlike a regular function that calculates and returns a single value (or a list/tuple of values) all at once, a generator yields values one at a time, only when requested (e.g., by a for loop or the next() function). This is what's meant by lazy—it only computes what is immediately needed.

State Suspension

- When a generator **yield**s a value, its entire state is frozen. This includes all local variables and the exact line of code it was on. When the next value is requested, the generator resumes execution from exactly where it left off, rather than starting over.

  Defining Generators:

 - A generator function is defined just like a normal function, but instead of using the return statement to send a final value back, it uses the yield statement to send a value back and pause execution.

 def generator_function_name(limit):

    """Generates numbers from 0 up to (but not including) limit."""

    n = 0

    while n < limit:

        yield n  # Pauses, sends 'n', and saves state

        n += 1   # Resumes from here on the next call

B. Generator Expressions

- A generator expression is a concise way to create a generator on a single line, similar to how list comprehensions are created, but using parentheses () instead of square brackets [].

List Comprehension (creates full list in memory)

list_comp = [x * x for x in range(10)]


Generator Expression (creates generator object, yields values lazily)

gen_exp = (x * x for x in range(10))   

Q.7  What are the advantages of using generators over regular functions?

- Generators are highly memory efficient because they only generate (or yield) one item at a time and hold the state needed to produce the next item.

- Generators: They don't store the entire sequence of results in memory. They use a fixed, small amount of memory, regardless of how large the sequence is. This is crucial for large datasets or infinite sequences.

- Regular Functions: If a regular function needs to return a large collection (like a list of a million numbers), it must first build that entire list in memory before returning it.

Lazy Evaluation:

- If you only need the first few elements of a potentially massive sequence, a generator will stop executing after yielding those few items.

- A regular function, on the other hand, will compute all results before you can access the first one. This can save significant computation time if not all results are ultimately needed.

Cleaner Code for Iterators:

- To manually create a custom iterator using a class, you have to define the special methods __iter__() and __next__() and handle the StopIteration exception.

- By using a generator function with the yield keyword, Python automatically handles all the complex iterator setup, state suspension, and exception raising behind the scenes.

Pipelining Data:

- This process flows naturally because each generator only yields an item when the next stage in the pipeline requests it. This avoids building intermediate lists, keeping the entire pipeline memory-efficient.

iterable: The sequence (list, tuple, etc.) whose elements will be passed to the function.

Q.10 What is the difference between `map()`, `reduce()`, and `filter()` functions in Python?

1. map(): Transformation

- The primary purpose of map() is transformation. It applies a function to every item in an iterable and returns a new iterable (a map object) of the results. The output sequence has the same number of items as the input sequence.

Goal: Change the value of items.

Operation: f(x)→y

Input: An iterable (e.g., [a, b, c])

Output: An iterable of transformed items (e.g., [f(a), f(b), f(c)])

Example: Squaring all numbers in a list.

numbers = [1, 2, 3]

squared = list(map(lambda x: x * x, numbers))

Result: [1, 4, 9]

2. filter(): Selection

The primary purpose of filter() is selection. It constructs an iterator (a filter object) from elements of an iterable for which a given function returns True. The output sequence typically has fewer or the same number of items as the input sequence.

Goal: Select items based on a condition.

Operation: Check if f(x) is True.

Input: An iterable (e.g., [a, b, c])

Output: An iterable of original items that satisfy the condition (e.g., [a, c] if f(b) was False)

Example: Selecting only even numbers from a list.

numbers = [1, 2, 3, 4]

evens = list(filter(lambda x: x % 2 == 0, numbers))

Result: [2, 4]

```python
# Q.1 Write a Python function that takes a list of numbers as input and returns the sum of all even numbers in the list.

def sum_of_even_numbers(numbers):
    total = 0
    for num in numbers:
        if num % 2 == 0:   # check if the number is even
            total += num
    return total

nums = [1, 2, 3, 4, 5, 6]
print("Sum of even numbers:", sum_of_even_numbers(nums))
```

```python
# Q.4  Write a Python function that checks if a given number is prime or not from 1 to 200.

def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

for num in range(1, 201):
    print(num, "is Prime" if is_prime(num) else "is Not Prime")
```

```python
# Q.5 Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of terms.

class FibonacciIterator:
    def __init__(self, n_terms):
        self.n_terms = n_terms
        self.count = 0
        self.a, self.b = 0, 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.n_terms:
            raise StopIteration

        if self.count == 0:
            self.count += 1
            return self.a
        elif self.count == 1:
            self.count += 1
            return self.b
        else:
            self.a, self.b = self.b, self.a + self.b
            self.count += 1
            return self.b


fib = FibonacciIterator(10)
for num in fib:
    print(num, end=" ")
```

```python
# Q.6 Write a generator function in Python that yields the powers of 2 up to a given exponent.

def powers_of_two(n):
    for exp in range(n + 1):
        yield 2 ** exp


for value in powers_of_two(5):
    print(value, end=" ")
```

```python
# Q.7 Implement a generator function that reads a file line by line and yields each line as a string.

def read_file_line_by_line(filename):
    try:
        with open(filename, "r") as file:
            for line in file:
                yield line.strip()
    except FileNotFoundError:
        print("Error: File not found.")

for line in read_file_line_by_line("example.txt"):
    print(line)
```

```python
# Q.9 Write a Python program that uses `map()` to convert a list of temperatures from Celsius to Fahrenheit.



celsius = [0, 10, 20, 30, 40]

fahrenheit = list(map(lambda c: (c * 9/5) + 32, celsius))

print("Celsius:", celsius)
print("Fahrenheit:", fahrenheit)
```

```python
# Q.10  Create a Python program that uses `filter()` to remove all the vowels from a given string.

def remove_vowels(text):
    vowels = "aeiouAEIOU"

    result = ''.join(filter(lambda ch: ch not in vowels, text))
    return result


s = "Hello World, Python is awesome!"
print("Original:", s)
print("Without vowels:", remove_vowels(s))
```
